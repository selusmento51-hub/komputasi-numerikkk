```python
import numpy as np
# Menyusun matriks koefisien A berukuran 3x3 dengan tipe data float (desimal)
# Tipe data float wajib digunakan agar pembagian elemen tidak dibulatkan menjadi integer
A = np.array([
    [3.0,  2.0, -1.0],
    [-1.0, 3.0,  2.0],
    [2.0, -1.0,  4.0]
], dtype=float)

# Menyusun vektor konstanta hasil B berukuran 3x1
B = np.array([10.0, 5.0, 12.0], dtype=float)

# Menampilkan matriks koefisien dan vektor hasil awal ke layar
print("--- Matriks Koefisien A ---")
print(A)
print("\n--- Vektor Hasil B ---")
print(B)


# %%
# ==============================================================================
# CELL 2: FUNGSI ALGORITMA ELIMINASI GAUSS-JORDAN
# ==============================================================================
def gauss_jordan_elimination(A_matrix, B_vector):
    """
    Menyelesaikan Sistem Persamaan Linear menggunakan metode Eliminasi Gauss-Jordan.
    Mengubah matriks augmented [A|B] menjadi bentuk matriks identitas tereduksi.
    """
    # Mengambil jumlah persamaan (n = 3) dari panjang vektor B
    n = len(B_vector)

    # Menggabungkan matriks A (3x3) dan vektor B (diubah ke kolom 3x1) menjadi matriks augmented (3x4)
    M = np.hstack([A_matrix, B_vector.reshape(-1, 1)])

    print("\n" + "="*45)
    print("Matriks Augmented Awal [A | B]:")
    print(M)
    print("="*45)

    # Iterasi utama untuk memproses setiap kolom diagonal utama (indeks i = 0, 1, 2)
    for i in range(n):
        # 1. PARTIAL PIVOTING:
        # Mencari indeks baris dengan nilai absolut terbesar pada kolom ke-i (mulai dari baris i ke bawah)
        # Tujuannya untuk mencegah pembagian dengan angka nol dan meminimalkan galat pembulatan
        max_row = i + np.argmax(np.abs(M[i:, i]))

        # Jika baris dengan nilai absolut terbesar bukan baris saat ini, tukar posisi kedua baris
        if i != max_row:
            M[[i, max_row]] = M[[max_row, i]]
            print(f"[Info] Menukar Baris {i+1} dengan Baris {max_row+1} (Partial Pivoting)")

        # 2. NORMALISASI BARIS PIVOT:
        # Mengambil nilai elemen pada diagonal utama M[i, i] sebagai pembagi (pivot)
        pivot = M[i, i]

        # Validasi: Jika elemen pivot bernilai 0, maka determinan matriks adalah 0 (tidak ada solusi unik)
        if pivot == 0:
            raise ValueError("Sistem persamaan tidak memiliki solusi unik (Matriks Singular).")

        # Membagi seluruh elemen pada baris i dengan nilai pivot sehingga diagonal utama M[i, i] menjadi 1
        M[i] = M[i] / pivot
        print(f"\n[Langkah {i+1}a] Normalisasi Baris {i+1} (dibagi dengan {pivot:.4f}):")
        print(np.round(M, 4))

        # 3. ELIMINASI KOLOM (ATAS & BAWAH PIVOT):
        # Melakukan perulangan ke seluruh baris j (kecuali baris pivot i) untuk mengenolkan elemen lain
        for j in range(n):
            if j != i:
                # Mengambil nilai pengali (faktor) pada kolom i baris ke-j yang ingin dijadikan 0
                faktor = M[j, i]

                # Mengurangi baris j dengan hasil perkalian faktor dan baris pivot i
                M[j] = M[j] - faktor * M[i]
                print(f"[Langkah {i+1}b] Eliminasi Baris {j+1}: Baris {j+1} - ({faktor:.4f} * Baris {i+1})")

        print("-" * 45)

    # Mengembalikan matriks augmented yang sisi kirinya sudah menjadi matriks identitas
    return M


# %%
# ==============================================================================
# CELL 3: EKSEKUSI PROGRAM & TAMPILAN SOLUSI AKHIR
# ==============================================================================
# Menggunakan blok try-except untuk menangani error jika matriks bersifat singular
try:
    # Memanggil fungsi Eliminasi Gauss-Jordan dengan input matriks A dan vektor B
    matriks_hasil = gauss_jordan_elimination(A, B)

    # Mengekstrak nilai solusi x, y, z dari kolom ke-4 (indeks 3) matriks hasil
    x = matriks_hasil[0, 3]
    y = matriks_hasil[1, 3]
    z = matriks_hasil[2, 3]

    # Menampilkan hasil akhir perhitungan dengan format 4 angka di belakang koma
    print("\n" + "="*45)
    print("HASIL AKHIR SOLUSI SISTEM PERSAMAAN LINEAR:")
    print("="*45)
    print(f"Nilai x = {x:.4f}")
    print(f"Nilai y = {y:.4f}")
    print(f"Nilai z = {z:.4f}")
    print("="*45)

except Exception as e:
    # Menampilkan pesan kesalahan apabila sistem gagal diselesaikan
    print(f"\nTerjadi kesalahan pada perhitungan: {e}")
```
